## 1. Configuração e Imports

```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from data_processing import load_and_parse_all

# Configuração de visualização
sns.set_style('whitegrid')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 100)

print("✅ Imports concluídos com sucesso!")
```

## 2. Carregamento e Parsing dos Dados

```python
# Carregar dados usando o módulo data_processing
df = load_and_parse_all(data_dir='../data/private')

print(f"\n📊 DataFrame carregado com sucesso!")
print(f"   Dimensões: {df.shape[0]} linhas × {df.shape[1]} colunas")
```

## 3. Inspeção Inicial dos Dados

```python
# Visualizar primeiras linhas
print("\n🔍 Primeiras 10 linhas do DataFrame:")
display(df.head(10))

# Informações sobre o DataFrame
print("\n📋 Informações do DataFrame:")
df.info()
```

```python
# Estatísticas descritivas
print("\n📈 Estatísticas Descritivas:")
print(f"   Total de cantos únicos: {df['canto'].nunique()}")
print(f"   Total de estrofes: {len(df)}")
print(f"   Canto mínimo: {df['canto'].min()}")
print(f"   Canto máximo: {df['canto'].max()}")
print(f"\n   Estrofes por canto (primeiros 5 cantos):")
for canto in range(1, 6):
    n_estrofes = len(df[df['canto'] == canto])
    print(f"      Canto {canto}: {n_estrofes} estrofes")
```

## 4. Análise de Valores Ausentes

```python
# Valores nulos e strings vazias
print("\n⚠️ Análise de Valores Ausentes:")
print("\nValores nulos:")
print(df.isnull().sum())

print("\nStrings vazias:")
empty_counts = {
    'italian': (df['italian'] == '').sum(),
    'william': (df['william'] == '').sum(),
    'john': (df['john'] == '').sum()
}
for col, count in empty_counts.items():
    print(f"   {col}: {count} strings vazias ({count/len(df)*100:.2f}%)")
```

## 5. Distribuição de Estrofes por Canto

```python
# Contar estrofes por canto para cada versão
stanza_counts = df.groupby('canto').agg({
    'italian': lambda x: (x != '').sum(),
    'william': lambda x: (x != '').sum(),
    'john': lambda x: (x != '').sum()
})

# Visualizar
fig, ax = plt.subplots(figsize=(14, 6))
stanza_counts.plot(kind='bar', ax=ax, color=['#2E86AB', '#A23B72', '#F18F01'])
ax.set_xlabel('Canto', fontsize=12)
ax.set_ylabel('Número de Estrofes', fontsize=12)
ax.set_title('Distribuição de Estrofes por Canto (Orlando Furioso)', fontsize=14, fontweight='bold')
ax.legend(['Italiano (Original)', 'William S. Rose', 'John Harington'])
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../outputs/figures/stanza_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n📊 Gráfico salvo em: outputs/figures/stanza_distribution.png")
```

## 6. Exemplos de Estrofes Parseadas

```python
# Exibir exemplo: Canto 1, Estrofe 1
print("\n📖 Exemplo: Canto 1, Estrofe 1")
print("="*80)

example = df[(df['canto'] == 1) & (df['estrofe'] == 1)].iloc[0]

print("\n🇮🇹 ITALIANO (Original):")
print(example['italian'][:300] + "..." if len(example['italian']) > 300 else example['italian'])

print("\n🇬🇧 WILLIAM STEWART ROSE:")
print(example['william'][:300] + "..." if len(example['william']) > 300 else example['william'])

print("\n🇬🇧 JOHN HARINGTON:")
print(example['john'][:300] + "..." if len(example['john']) > 300 else example['john'])
```
